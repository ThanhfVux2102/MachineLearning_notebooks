```python
import tensorflow as tf
import tensorflow_datasets as tfds
import numpy as np
import matplotlib.pyplot as plt

```

```python
# import data and label

(ds_train,ds_test),ds_info = tfds.load('fashion_mnist',split=['train','test'],shuffle_files = True,with_info = True  )
```

```python
#preprocessing

def ds_normalize (elements):
  image = elements['image']
  labels = elements['label']
  return tf.cast(image,tf.float32)/255, labels

ds_train = ds_train.map(ds_normalize,num_parallel_calls = tf.data.AUTOTUNE)
ds_train = ds_train.cache()
ds_train = ds_train.shuffle(ds_info.splits['train'].num_examples,seed = 42 )
ds_train = ds_train.batch(128)
ds_train = ds_train.prefetch(tf.data.AUTOTUNE)
```
